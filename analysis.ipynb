```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# Load CSV file
df = pd.read_csv("../survey_data.csv")

# Show first rows
df.head()
```

```python
# Convert income to numeric
df["income"] = pd.to_numeric(df["income"])

# Create chart
plt.figure(figsize=(8,5))

sns.barplot(x="age", y="income", data=df)

plt.title("Age vs Income")
plt.xlabel("Age")
plt.ylabel("Income")

# Save chart
plt.savefig("age_income_chart.png")

plt.show()
```

```python
# Create spending totals
spending_data = {
    "Utilities": pd.to_numeric(df["utilities_amount"]).sum(),
    "Entertainment": pd.to_numeric(df["entertainment_amount"]).sum(),
    "School": pd.to_numeric(df["school_amount"]).sum(),
    "Shopping": pd.to_numeric(df["shopping_amount"]).sum(),
    "Healthcare": pd.to_numeric(df["healthcare_amount"]).sum()
}

# Create dataframe
spending_df = pd.DataFrame(
    list(spending_data.items()),
    columns=["Category", "Amount"]
)

# Plot chart
plt.figure(figsize=(8,5))

sns.barplot(x="Category", y="Amount", data=spending_df)

plt.title("Spending Categories")
plt.xlabel("Category")
plt.ylabel("Amount")

# Save chart
plt.savefig("spending_categories_chart.png")

plt.show()
```
